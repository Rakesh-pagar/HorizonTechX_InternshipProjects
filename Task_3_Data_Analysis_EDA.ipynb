```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Rohan",
             "Neha", "Akash", "Pooja", "Vikas", "Anjali"],
    "Marks": [78, 92, 65, 88, 55, 74, 81, 95, 62, 85],
    "Attendance": [85, 95, 70, 90, 65, 80, 88, 96, 68, 92]
}
df = pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Marks"])

plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.title("Student Marks")

plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Attendance"])
plt.xlabel("Student Name")
plt.ylabel("Attendance (%)")
plt.title("Student Attendance")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Attendance"], df["Marks"])
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.title("Relationship Between Attendance and Marks")

plt.show()
```

```python
correlation = df["Attendance"].corr(df["Marks"])
print("Correlation between Attendance and Marks:", correlation)
```

```python
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)

df["Performance"] = np.where(
    df["Marks"] >= average_marks,
    "Above Average",
    "Below Average"
)
df
```

```python
df.isnull().sum()
```

```python
df["Performance"].value_counts()
```

```python
performance_count = df["Performance"].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(performance_count.index, performance_count.values)
plt.xlabel("Performance")
plt.ylabel("Number of Students")
plt.title("Student Performance Distribution")

plt.show()
```

```python
average_marks = df["Marks"].mean()
average_attendance = df["Attendance"].mean()
highest_marks = df["Marks"].max()
lowest_marks = df["Marks"].min()
above_average = (df["Performance"] == "Above Average").sum()
below_average = (df["Performance"] == "Below Average").sum()
correlation = df["Attendance"].corr(df["Marks"])

print("========== STUDENT PERFORMANCE ANALYSIS REPORT ==========")
print()
print("Total Students:", len(df))
print("Average Marks:", average_marks)
print("Average Attendance:", average_attendance, "%")
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
print("Above Average Students:", above_average)
print("Below Average Students:", below_average)
print("Attendance-Marks Correlation:", round(correlation, 3))
print()
print("========== KEY INSIGHTS ==========")
print("1. 6 out of 10 students are above average.")
print("2. 4 out of 10 students are below average.")
print("3. The highest marks are 95 and the lowest marks are 55.")
print("4. The dataset contains no missing values.")
print("5. A strong positive relationship is observed between attendance and marks.")
```
