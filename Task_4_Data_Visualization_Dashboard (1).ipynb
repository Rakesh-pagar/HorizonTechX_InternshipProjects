```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Name":["Amit", "Sneha", "Rahul", "Priya", "Rohan",
             "Neha", "Akash", "Pooja", "Vikas", "Anjali"],
    "Marks":[78, 92, 65, 88, 55, 74, 81, 95, 62, 85],
    "Attendance": [85, 95, 70, 90, 65, 80, 88, 96, 68, 92]
}
df = pd.DataFrame(data)
average_marks = df["Marks"].mean()
df["Performance"] = np.where(
    df["Marks"] >= average_marks,
    "Above Average",
    "Below Average"
)
print("Dataset loaded successfully!")
df
```

```python
fig = plt.figure(figsize=(14, 11))

# KPI 1 Total Students
ax1 = plt.subplot(3, 2, 1)
ax1.text(0.5, 0.5, f"{len(df)}\nTotal Students",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax1.axis("off")

# KPI 2 Average Marks
ax2 = plt.subplot(3, 2, 2)
ax2.text(0.5, 0.5, f"{df['Marks'].mean():.1f}\nAverage Marks",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax2.axis("off")

# KPI 3 Average Attendance
ax3 = plt.subplot(3, 2, 3)
ax3.text(0.5, 0.5, f"{df['Attendance'].mean():.1f}%\nAverage Attendance",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax3.axis("off")

# KPI 4 Highest Marks
ax4 = plt.subplot(3, 2, 4)
ax4.text(0.5, 0.5, f"{df['Marks'].max()}\nHighest Marks",
         ha="center", va="center", fontsize=18, fontweight="bold")
ax4.axis("off")

# Marks Graph
ax5 = plt.subplot(3, 2, 5)
ax5.bar(df["Name"], df["Marks"])
ax5.set_title("Student Marks")
ax5.set_xlabel("Student")
ax5.set_ylabel("Marks")
ax5.tick_params(axis="x", rotation=45)

# Attendance Graph
ax6 = plt.subplot(3, 2, 6)
ax6.bar(df["Name"], df["Attendance"])
ax6.set_title("Student Attendance")
ax6.set_xlabel("Student")
ax6.set_ylabel("Attendance (%)")
ax6.tick_params(axis="x", rotation=45)

plt.suptitle("Student Performance Dashboard", fontsize=20, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 12))

plt.subplot(4, 2, 1)
plt.text(0.5, 0.5, f"{len(df)}\nTotal Students",
         ha="center", va="center", fontsize=18, fontweight="bold")
plt.axis("off")

plt.subplot(4, 2, 2)
plt.text(0.5, 0.5, f"{df['Marks'].mean():.1f}\nAverage Marks",
         ha="center", va="center", fontsize=18, fontweight="bold")
plt.axis("off")

plt.subplot(4, 2, 3)
plt.text(0.5, 0.5, f"{df['Attendance'].mean():.1f}%\nAverage Attendance",
         ha="center", va="center", fontsize=18, fontweight="bold")
plt.axis("off")

plt.subplot(4, 2, 4)
plt.text(0.5, 0.5, f"{df['Marks'].max()}\nHighest Marks",
         ha="center", va="center", fontsize=18, fontweight="bold")
plt.axis("off")

# GRAPH 1 
plt.subplot(4, 2, 5)
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.xticks(rotation=45)


# GRAPH 2
plt.subplot(4, 2, 6)
plt.bar(df["Name"], df["Attendance"])
plt.title("Student Attendance")
plt.xlabel("Student")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=45)

#  GRAPH 3 
plt.subplot(4, 2, 7)
plt.scatter(df["Attendance"], df["Marks"])
plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

# GRAPH 4 
plt.subplot(4, 2, 8)
performance_count = df["Performance"].value_counts()
plt.bar(performance_count.index, performance_count.values)
plt.title("Performance Distribution")
plt.xlabel("Performance")
plt.ylabel("Number of Students")


# TITLE
plt.suptitle(
    "Student Performance Dashboard",
    fontsize=20,
    fontweight="bold"
)
plt.tight_layout()
plt.show()
```
